```python
import os
import pandas as pd
import numpy as np

# directory containing experiment folders
base_path = "./data"

# subdirectory for a single run of the experiment 
# please adjust the following string to suite the experiment name
experiment_path = "experiment_dummy"
data_path = os.path.join(base_path, experiment_path)
print(f"Loading data from: {data_path}")

# data structure for all experiments
experiments = {}

# iterate over all subfolders in data_path
folders = sorted([f for f in os.listdir(data_path) if os.path.isdir(os.path.join(data_path, f))])

for folder in folders:
    folder_path = os.path.join(data_path, folder)
    
    # helper function to read
    def load_csv(filename):
        return pd.read_csv(os.path.join(folder_path, filename), header=None).values
    
    experiments[folder] = {
        "translation": load_csv("translation_vector.csv"),
        "quaternion": load_csv("rotation_quaternion.csv"),
    }

# check loaded data
first = folders[0]
print("Example experiment:", first)
print("Translation:\n", experiments[first]["translation"])
print("Quaternion:\n", experiments[first]["quaternion"])
```

```python
# list of translation and rotation deltas
trans_list = []
rot_list = []

# populate lists with deltas
for experiment in experiments:
    trans_list.append(experiments[experiment]["translation"])
    rot_list.append(experiments[experiment]["quaternion"])

# check content of lists
print("Translations:", trans_list)
print("Rotations:", rot_list)
```

```python
# compute mean and stddev for translations
trans_mean = np.mean(trans_list, axis=0)
rot_mean = np.mean(rot_list, axis=0)

# compute stddev for translations
# trans_stddev = st.stdev(trans_list)
# rot_stddev = st.stdev(rot_list)

# check results
print("Mean Translation:\n", trans_mean)
print("Mean Quaternion:\n", rot_mean)
# print("Translation StdDev:\n", trans_stddev)
# print("Quaternion StdDev:\n", rot_stddev)
```
